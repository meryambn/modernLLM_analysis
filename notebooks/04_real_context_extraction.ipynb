cell 4 - Build a Context Extraction Function

```python
def extract_occurrences(
    target_word,
    max_examples=100
):

    contexts = []

    for sentence in semcor.sents():

        words = [
            word.lower()
            for word in sentence
        ]

        if target_word.lower() in words:

            contexts.append(
                " ".join(sentence)
            )

        if len(contexts) >= max_examples:
            break

    return contexts
```
